```python
# ----------------------------------------------------------
# 1. Bar Chart: How Discounts Impact Business Profits
# ----------------------------------------------------------

import matplotlib.pyplot as plt
import seaborn as sns

# Data
categories = [
    "Decreased Profit",
    "Boost Short-Term Sales",
    "No Major Impact",
    "Increased Revenue"
]

values = [60, 25, 10, 5]

# Create Figure
plt.figure(figsize=(8, 5))

# Plot
sns.barplot(x=categories, y=values, palette="Reds_d")

# Labels
plt.title("How Discounts Impact Business Profits", fontsize=14)
plt.xlabel("Seller Perception of Discounts")
plt.ylabel("Percentage (%)")
plt.ylim(0, 70)
plt.xticks(rotation=20)

# Show Plot
plt.show()
```

```python
# ----------------------------------------------------------
# 2. Pie Chart: Seller Challenges Due to Discounting
# ----------------------------------------------------------

import matplotlib.pyplot as plt

# Data
labels = [
    "Reduced Profit Margins",
    "Customers Expect Frequent Discounts",
    "Brand Devaluation",
    "No Challenges"
]

sizes = [50, 30, 15, 5]

colors = [
    "#ff9999",
    "#66b3ff",
    "#99ff99",
    "#ffcc99"
]

# Create Figure
plt.figure(figsize=(6, 6))

# Plot
plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct="%1.1f%%",
    startangle=140
)

# Title
plt.title("Challenges Faced by Businesses Due to Discounts")

# Show Plot
plt.show()
```
